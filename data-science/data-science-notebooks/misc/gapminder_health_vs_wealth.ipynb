```python
def getsize(pop):
  if pop<= 9999999:
    return "1 Million"
  elif pop <= 99999999:    # 10 million or below
        return '10 Million'
  elif pop <= 999999999:   # 100 million or below
        return '100 Million'
  else:                    # Larger than 100 million
        return '1000 Million'
```

```python
def value_to_float(x):
    if type(x) == float or type(x) == int:
        return x
    if 'K' in x:
        if len(x) > 1:
            return float(x.replace('K', '')) * 1000
        return 1000.0
    if 'M' in x:
        if len(x) > 1:
            return float(x.replace('M', '')) * 1000000
        return 1000000.0
    if 'B' in x:
        return float(x.replace('B', '')) * 1000000000
    return 0.0
```
