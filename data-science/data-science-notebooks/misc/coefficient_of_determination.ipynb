```python
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
f=np.array([100,95,90,102,94])
f = np.random.randint(low=1, high=102, size = 10)


```

```python
c= (f-32)*5/9
c
```

```python
dfT = pd.DataFrame({'F': f, 'C': c})
dfT
```

```python
plt.scatter (dfT.F, dfT.C)
```

```python
import sklearn.linear_model as skl
model = skl.LinearRegression()
x= dfT.F.values.reshape(-1,1)
model.fit(x, dfT.C)
y_pred = model.predict(x)
plt.scatter(x, dfT.C, marker ='*', s= 600)
plt.plot(x, y_pred, "r-o")
model.score(x,dfT.C)
model.coef_[0]
```

```python
dfT[['F', 'C']].corr()
```
