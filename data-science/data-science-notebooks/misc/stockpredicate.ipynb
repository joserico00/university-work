```python
import pandas as pd
import numpy as np
import datetime

# Create an empty list to store transaction data
transaction_data = []
```

```python
def simulate_transactions(duration):
    start_time = datetime.datetime.now()
    prices = []
    
    while (datetime.datetime.now() - start_time).seconds < duration:
        new_transaction = {
            "item_id": np.random.randint(1, 100),
            "price": np.random.uniform(5, 500),
        }
        transaction_data.append(new_transaction)
        prices.append(new_transaction["price"])

        # Calculate and print the median price
        prices.sort()
        n = len(prices)
        median_price = prices[n//2] if n % 2 else (prices[n//2 - 1] + prices[n//2]) / 2
        print("Current median price:", median_price)

        #datetime.timedelta(seconds=1).sleep()  # Sleep for 1 second before the next iteration
```

```python
simulate_transactions(10)
```
