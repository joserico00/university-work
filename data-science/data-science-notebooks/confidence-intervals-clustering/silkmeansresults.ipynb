```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score





```

```python

raw_data = make_blobs(n_samples=200, n_features=2, centers=3, cluster_std=1.5)
```

```python
model = KMeans(n_clusters=3)
model.fit(raw_data[0])
```

```python
silh_score = silhouette_score(raw_data[0], model.labels_)
print("For n_clusters =", 3, "The average silhouette_score is :", silh_score)
```

```python
plt.scatter(raw_data[0][:, 0], raw_data[0][:, 1], c=model.labels_)
plt.scatter(model.cluster_centers_[:, 0], model.cluster_centers_[:, 1], c='red', marker='x')
plt.show()
```

```python


silh_scores = []
for k in range(2, 10):
    model = KMeans(n_clusters=k)
    model.fit(raw_data[0])
    silh_score = silhouette_score(raw_data[0], model.labels_)
    silh_scores.append(silh_score)

plt.plot(range(2, 10), silh_scores, 'go-', linewidth=3, markersize=12)
plt.xlabel("Number of clusters")
plt.ylabel("Average silhouette score")
plt.show()
```

# comparacion entre k-means and silhouette

### Son similares en el sentido que ambos son utilizado en el contexto de clustering, kmeans es un algoritmo que hace el clustering y silhouette scoring es la manera de evaluar los resultados de clustering



### son diferente porque k-means es un algoritmo que agrupa data, mientras silhouette score es usado para medir cuan cerca son los puntos de un cluster del un otro cluster
