Hasta aquí la limpieza. Vamos a hacer algo de "wrangling". El objetivo final es identificar todos los vuelos que sales de San Juan y llegan a Tucson. Si no hay vuelos directos, que tengan solamente una escala. Recuerde que toda la data está guerdada como objetos ('strings') de acuerdo a la información anterior.

```python
# Este slice me identifica todos los vuelos procedentes de SJ
df[df['Origin Airport'] == 'SJU']
```

Si quiero identificar si algún valor se encuentra en alguna de las columnas, puedo usar el objeto "in" de Python. Por ejemplo:

```python
'321' in df['Aircraft Type'].values
# si tengo aviones de tipo '321'. Debo usar el atributo "values" porque de lo contrario busca en los índices (Python, no Pandas).
```

```python
# Puedo buscar todos los vuelos que salen de SJ pero que no son en aviones de tipo 321:
df[(df['Origin Airport'] == 'SJU') & (df['Aircraft Type'] != '321')]
```

Si en lugar de comparar con un valor quisiera comparar con o buscar en varios, puedo usar el método "isin" de Pandas que itera un objeto. Por ejemplo, voy a hacer una lista de varios tipos de aviones y verificar cuáles vuelos desde SJ usan esos aviones:

```python
notAllowed = ['319', '320', '321', '321E', '321H', '321S', '321X', '32Q','32B']
```

```python
df[(df['Origin Airport'] == 'SJU') & (df['Aircraft Type'].isin(notAllowed))]
# Primero busco que aviones salen de SJ y hago la intersección ("and" es intersección) con todos los aviones en la lista
```

```python
# Si, por el contrario, quisiera todos los vuelos que NO usan el tipo de avion en la lista, uso el complemento
# (recuerde que la negación es complemento). Para negar se puede usar tanto ! como ~
df[~df['Aircraft Type'].isin(notAllowed)]
```

A continuavión, todos los vuelos que salen de SJ pero usando aviones que no están en la lista anterior:

```python
dfd= df[(df['Origin Airport'] == 'SJU') & (~df['Aircraft Type'].isin(notAllowed))]
dfd
```

# Asignación

Busque todos los vuelos que salen de SJ y llegan a Tucson (código: TUS). Si no hay vuelos directos, busque vuelos con máximo de una escala. Ninguno de los vuelos (sea directo o alguno de los tramos) debe usar aviones en la lista "notAllowed".

```python
#df[(df['Origin Airport'] == 'SJU') & (df['Dest. Airport'] == 'TUS') | ()  ]
```

```python
dfd=df[(df['Origin Airport'] == 'SJU') & (~df['Aircraft Type'].isin(notAllowed))]
dfd
```

```python
dfa= df[(df['Dest. Airport'] == 'TUS') & (~df['Aircraft Type'].isin(notAllowed))]
dfa
```

```python
dfD=dfa[dfa['Origin Airport'].isin(dfd['Dest. Airport'])].reset_index(drop=True)
dfD
```

```python
#dfA= dfD[dfD['Dest. Airport'].isin(dfa[])]
```

```python
dfA= dfd[dfd['Dest. Airport'].isin(dfD['Origin Airport'])].reset_index(drop=True)
dfA
```

```python
result = pd.concat([dfA,dfD])
result
```

```python
newDF= dfD.merge(dfA,how='left',left_on='Origin Airport',right_on='Dest. Airport')
#newDF= dfD.merge(dfA,how='left',left_on='Dest. Airport',right_on='Origin Airport')
newDF
```
