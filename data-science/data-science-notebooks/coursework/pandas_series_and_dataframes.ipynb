# Pandas Data Structures

**Pandas** es una librería escrita con el propósito de manipular y analizar data usando Python (aunque no está escrita solamente en Python).

Usamos Pandas en conjunto con **Numpy**, otra librería escrita con el propósito de maniplar **data numérica** en Python. Es necesario usar Numpy con Pandas porque Pandas está escrita sobre Numpy, aunque a veces causan confusión los métodos que ambas usan pues los propósitos parecen los mismos.

Pandas usa dos tipos básicos de estructuras para organizar y analizar la data: **Series** (vectores de una dimensión) y **Dataframes** (tablas en dos dimensiones).

```python
import numpy as np
import pandas as pd
```

## Pandas Series

Las series son arreglos unidimensionales estructurados. Es decir, además de la data, contienen indexación para acceder y organizar la misma.

Nuestra serie puede ser un arreglo creado con data numérica o con data de tipo mixto. Para convertir el arreglo a una serie, usamos el método **pd.Series(data)** en donde data es el nombre del objeto conteniento el arreglo. 

```python
data1 = [1,2,3,4, "hola"] #esto es una "lista" en Python. Listas y arreglos van entre [ ]
```

```python
dataS = pd.Series(data1) #el método devuelve la estructura, por lo tanto, debemos asignarlo auna variable
dataS

# Note como el tipo de data es "object". Sin embargo, si cambia la lista a una numérica solamente, 
# el tipo será int64. Note, además, la indexación, la cuál comienza en 0.
```

```python
dataS.index 
# el atributo "index" puede usarse para desplegar donde comienza, termina y como cambian los índices en una 
# estructura. Por defecto comienzan en 0. También puede usarse para modificar dichos índices.
```

Podemos cambiar los índices de diferentes maneras. Si quisiéamos que comienzen, por ejemplo, en 1 en ligar de 0, lo mas simple es sumándole 1 a los índices existentes:

*Nota: si corre varias veces el código, cambiará el índice cada vez que lo corra a menos que haya creado nuevamente la serie*

```python
dataS.index +=1
dataS
```

Pero también podemos elegir en donde empezar, terminar y de cuanto en cuanto cambiar usando el método de Pandas RangeIndex. Como quien debe invocar el método es Pandas y no la estructura, este método debe usarse cuando se crea la estructura (serie o dataframe):

```python
dataIndexed = pd.Series(data1, pd.RangeIndex(start=4, stop=24, step=4)) #recuerde que el valor de stop no está incluido
dataIndexed
```

También podemos usar data no numérica (strings) para los índeices:

```python
dataS.index = ['primero', 'segundo', 'tercero', 'cuarto', 'quinto']
dataS
```

A continuación, un ejemplo de porqué es preferible usar Numpy que los métodos nativos de Python para operacones matemáticas computacionalmente intensas. Comparamos la suma nativa de Python con la de Numpy en un vector random:

```python
dataR = np.random.rand(10000)
```

```python
%timeit sum(dataR)
%timeit np.sum(dataR)
```

## Pandas Data Frames

Los Dataframes son estructuras tabulares (o, simplemente, tablas) que al igual que las Series también están indexadas. Podemos construir dataframes de diversas maneras, siempre con el método de Pandas (Pandas es quien invoca) DataFrame(). En la clase anterior, vimos como crear un dataframe de una tabla guardada en un archivo CSV. A continuación, como crear un dataframe de un diccionario en Python:

```python
data2 = {'nombre': ['Luis', 'Miguel','Pedro'],
        'id': ['801000000', '801666666','55555555'],
        'gpa': [3.80, 4.00,3.35]}

data2
```

```python
dataF = pd.DataFrame(data2)
dataF
# Esta manera de crear un dataframe, sin embargo, no es práctica para grandes cantidades de data.
```

También podemos crear un dataframe de data numérica solamente que ha sido guardada en un archivo. Primero, vamos a generar una matriz random usando Numpy y luego vamos a guardarla en un archivo CSV. 

```python
matx = np.random.rand(5,8)
matx
```

La instrucción de grabar **NO** devuelve nada y por lo tanto no hay que asignarla a ninguna variable.
Quien está guardando es Numpy, no Pandas. Para grabar con Pandas, quien invoca es el dataframe y aún no lo hemos creado.
**Si no especifa lo contario, Numpy usará con esta instrucción espacios en blanco como separadores (delimiter = ' '). Si quiere que guarde con comas, debe especificarlo.**

```python
np.savetxt('ranDat.csv', matx) #(dirección, data) o (dónde, qué)
# np.savetxt('ranDat.csv', matx, delimiter = ',')
```

Ahora vamos a leer el archivo (que puede estar creado de alguna ocasión anterior). Como la lectura **devuelve** el objeto que manipula la data, debe asignarlo a una variable. Primero, vamos a leer con Numpy: 

```python
data3 = np.loadtxt('ranDat.csv')
data3
```

Aunque está leyendo un CSV, la instrucción lo lee sin comas porque así es por defecto "loadtxt()". Si el archivo estuviera guardado usando comas, la instrucción anterior habría dado un error. En ese caso, debe añadir **delimiter = ','** a la instrucción de lectura.

Ahora, vamos a leer usando Pandas: 

```python
data3 = pd.read_csv('ranDat.csv')
data3
```

Note como al leer con Pandas, se devuelve ya un dataframe creado. Esto puede ser problemático cuando la data es solamente numérica, como en este caso, pues no hay nombres de las columnas. 

Note, además, que guardámos con la instrucción de Numpy (y aunque dijimos CSV, el default en delimitadores es ' '). Así que, cuando leamos con Pandas (que espera leer un CSV), la data se leerá como una sola columna pues no hay separadores.

Aunque podemos leer especificando que la data está separada por espacios usando **data3 = pd.read_csv('ranDat.csv', sep = ' ')** (aunque en Pandas el parámetro default para los separadores es "sep", también acepta "delimiter" como un alias, se leerá como un dataframe de cierta cantidad de líneas y **los encabezados de las columnas será la primera línea de data**. Por Lo tanto, al intentar luego hacer el dataframe, tendremos errores ya que intentaremos sobrrescribir los encabezados y añadir una línea de manera incorrecta: 

```python
data3 = pd.read_csv('ranDat.csv', sep = ' ')
data3
```

```python
dataF2 = pd.DataFrame(data3,columns=['largo', 'ancho', 'prof', 'alt','masa', 'radio', 'vol','dens'])
dataF2
```

Por lo tanto, y como explicado en una clase anterior, data numérica solamente debe ser leida usando Numpy:

```python
data3 = np.loadtxt('ranDat.csv')
dataF2 = pd.DataFrame(data3,columns=['largo', 'ancho', 'prof', 'alt','masa', 'radio', 'vol','dens'])
dataF2
```

Vamos ahora a añadir una columna. Esta puede ser añadida como data manual, leida de otra parte o como el resultado de alguna operación con otras columnas del dataframe. Simplenmente, creamos la otra columna como un vector del dataframe con el nombre que deseamos y le asignamos los valores correspondientes:

```python
dataF2['area']= dataF2['largo']*dataF2['ancho']
dataF2
```

Podemos "reorganizar" las columnas en otro orden, si así deseamos. Con el método siguiente, en realidad estamos creando nuevamente el dataframe (note que omite la columna de área que se añadió). Una manera mas práctica (pero mucho costosa y poco eficiente) es la segunda opción que se presenta. Mas adelante veremos maneras mas eficientes:

```python
# dataF2 = pd.DataFrame(data3,columns=['prof', 'alt','masa', 'radio', 'vol','dens','largo', 'ancho' ])
# Luego de ejecutar esta instrucción, debe repetir las de arriba (crear con data3 y añadir area) pues perderá la columna de area 
```

```python
# dataF2 = pd.DataFrame(data3,columns=['prof', 'alt','masa', 'radio', 'vol','dens','largo', 'ancho' ])
dataF2 = dataF2 [['prof', 'alt','masa', 'radio', 'vol','dens','largo', 'ancho', 'area']]
dataF2
```

Para añadir una fila, debemos primero crearla como un diccionario, y entonces usamos "append". Sin embargo, cuando se anejan diccionarios en dataframes, hay que especificar que se "ignore el índice":

```python
newDat = {'alt': 12, 'radio':14, 'vol':15,'dens':16,'largo':17, 'masa':13,'ancho':18, 'area': 25,'prof': 11}
dataF2.append(newDat, ignore_index=True)
```

Como en todo diccionario, note que el orden en que crea la data no importa y lo asigna donde corresponde
**TIENE que asignar** de nuevo al dataframe o pierde lo que anejo. 

```python
dataF2
```

```python
dataF2 = dataF2.append(newDat, ignore_index=True)
dataF2
# Cada vez que corra esta instrucción anejará otra fila
```

## Ordenamiento

Para ordenar, debemos indicar que criterio se usará (nombre de la columna). Quien invoca el método es el dataframe. Por defecto, se ordena de manera ascendente; si queremos ordenar de manera descendente, usamos "ascending = False" como argumento.

```python
dataF.sort_values(by ='nombre', ascending = False)
# Si no asigna, pierde la data ordenada.
```
