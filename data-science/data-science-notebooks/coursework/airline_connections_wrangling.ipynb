En esta parte vamos a practicar algunas instrucciones de limpieza de datos y jugar con algunas instrucciones de modificación y manipulación.

Comenzamos por importar las librerías y leer el archivo.

```python
import pandas as pd
import numpy as np
import datetime
```

```python
df = pd.read_csv('datasets/ConfirmedFSMarch2023.csv', skiprows = [0,1,2,3,6,7],skipfooter = 2272, engine = 'python')
```

```python
# Como hay dos columnas que se llman igual, Pandas indexa la segunda para distinguirla (lo aprendí con ustedes). De ahí el .1 luego de Flight
df.rename({'Flight.1': 'Flight'},axis = 1, inplace = True)

# Si quiero cambiar nombres individuales a columnas uso un "diccionario", especificando el eje. Recuerde que si no asigno los cambios, debo hacerlos "in place" o los perderé.
```

```python
# Quiero que el header sea la union de el header mas la primera fila. Uso iloc para asegurarme que es la primera fila, no la fila con label 0.
df.columns = df.columns + ' ' + df.iloc[0]
```

```python
df = df.drop([0]).reset_index(drop=True)
df
```

```python
notAllowed = ['319', '320', '321', '321E', '321H', '321S', '321X', '32Q','32B']
```

```python
df['Departure Time'] = pd.to_datetime(df['Departure Time'],format= '%H:%M')
df['Arrival Time'] = pd.to_datetime(df['Arrival Time'],format= '%H:%M')
```

```python
# También puede hacer parsing al string de tiempo (no puede correr el de arriba porque da error ya que ya 
# convirtió a datetime y no puede hacer split si no es string - ese es el error que estaba teniendo en clases)

# str.split(expand = True) separa el string en columnas de acuerdo al criterio de split. Esto es lo que les quería mostrar:
df['Departure Time'].str.split(':',expand = True)
```

```python
df['Departure Time'] = pd.to_timedelta((df['Departure Time'].str.split(':',expand = True).astype(int)*(60,1)).sum(axis=1),unit='min')
df['Arrival Time'] = pd.to_timedelta((df['Arrival Time'].str.split(':',expand = True).astype(int)*(60,1)).sum(axis=1),unit='min')
df
```

```python
dfSJ = df[(df['Origin Airport'] == 'SJU') & (~df['Aircraft Type'].isin(notAllowed))]
dfSJ
```

```python
dfTUS = df[(df['Dest. Airport'] == 'TUS') & (~df['Aircraft Type'].isin(notAllowed))]
dfTUS
```

```python
connect = list(set(dfTUS['Origin Airport']).intersection(set(dfSJ['Dest. Airport'])))
connect
# connect = list(set(dfTUS['Origin Airport']) & (set(dfSJ['Dest. Airport'])))
```

```python
dfSJ_ = df[((df['Origin Airport'] == 'SJU') & (df['Dest. Airport'].isin(connect)) & (~df['Aircraft Type'].isin(notAllowed)))].reset_index(drop=True) 
dfSJ_ 
```

```python
_dfTUS = df[((df['Dest. Airport'] == 'TUS') & (df['Origin Airport'].isin(connect)) & (~df['Aircraft Type'].isin(notAllowed)))].reset_index(drop=True)
_dfTUS
```

```python
df[((df['Origin Airport'] == 'SJU') & (df['Dest. Airport'].isin(connect)) & (~df['Aircraft Type'].isin(notAllowed))) | 
((df['Dest. Airport'] == 'TUS') & (df['Origin Airport'].isin(connect)) & (~df['Aircraft Type'].isin(notAllowed)))]

# (df[((df['Origin Airport'] == 'SJU') & (df['Dest. Airport'].isin(connect))) | 
# ((df['Dest. Airport'] == 'TUS') & (df['Origin Airport'].isin(connect)))]) 
```

```python
lenSJ = len(dfSJ_.index)
lenTUS = len(_dfTUS.index)
```

```python
if lenSJ < lenTUS:
    filler1 = pd.Series([None]*(lenTUS - lenSJ))
    connectT1 = pd.concat([dfSJ_['Arrival Time'],filler1],ignore_index=True)
    connectT2 = _dfTUS['Departure Time']
elif lenSJ > lenTUS:
    filler2 = pd.Series([None]*(lenSJ - lenTUS))
    connectT2 = pd.concat([_dfTUS['Departure Time'],filler2],ignore_index=True)
    connectT1 = dfSJ_['Arrival Time']
else:
    connectT1 = dfSJ_['Arrival Time']
    connectT2 = _dfTUS['Departure Time']

connectT1
# connectT2
```

```python
minConexTime = pd.Timedelta(hours = 2)
```

```python
dfD = dfSJ_.loc[connectT1 < (connectT2 + minConexTime)]
dfD
```

```python
dfA = _dfTUS.loc[connectT1 < (connectT2 + minConexTime)]
dfA
```

```python
newDF = dfD.merge(dfA, how= 'cross')
newDF.loc[(newDF['Dest. Airport_x'] == newDF['Origin Airport_y'])]
# newDF
```

```python
newDF = dfD.merge(dfA, how= 'left', left_on='Dest. Airport',right_on='Origin Airport')
newDF
```

```python
newDF = dfD.merge(dfA, how= 'left', on='Effective Date')
newDF
```

```python
newDF = dfD.merge(dfA, how= 'right', left_on='Arrival Time',right_on='Departure Time')
newDF
```
